## tensorの順番をランダムに並びかえる

```python
import torch

# サンプルの[N, M]テンソル
N, M = 5, 3
tensor = torch.arange(N * M).view(N, M)
print("元のテンソル:")
print(tensor)

# ランダムなインデックスを生成
shuffled_indices = torch.randperm(N)

# 行を並び替える
shuffled_tensor = tensor[shuffled_indices]
print("\nシャッフル後のテンソル:")
print(shuffled_tensor)
```
