```python
import torch
import torch.nn as nn

class YOLOFeatureExtractor(nn.Module):
    def __init__(self, model: nn.Module):
        super().__init__()
        self.model = model

        # 返したいレイヤーの名前を指定します
        self.return_layers = {
            "model.0": "0",  # 最初のスケール
            "model.1": "1",  # 中間スケール
            "model.2": "2",  # 最も小さいスケール
        }

    def forward(self, x):
        out = {}
        for name, layer in self.model.named_children():
            x = layer(x)
            if name in self.return_layers:
                out[self.return_layers[name]] = x
        return out

# YOLOv8 モデルを読み込む（例：ultralytics ライブラリを使用）
from ultralytics import YOLO

# YOLOv8 モデルをインスタンス化
yolo_model = YOLO('yolov8n.pt')  # 'yolov8n.pt' はYOLOv8のモデルファイル名

# 特徴マップを取得するためのモデルを作成
feature_extractor = YOLOFeatureExtractor(yolo_model.model)

# 入力テンソルの例
input_tensor = torch.randn(1, 3, 640, 640)  # バッチサイズ1, チャンネル数3, 640x640 画像

# 特徴マップを取得
features = feature_extractor(input_tensor)

# 各特徴マップのサイズを確認
for name, feature in features.items():
    print(f"Feature {name} shape: {feature.shape}")
```
