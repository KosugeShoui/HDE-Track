```python
import cv2
import matplotlib.pyplot as plt

# 動画ファイルのパス
#video_path = 'w_output_video_jet/visem_src_notime_results.mp4'
#video_path = 'w_output_video_jet/visem_time_weight_results.mp4'
video_path = 'w_output_video_jet/input_visem_time_weight_results.mp4'
#video_path = 'w_output_video_jet/input_visem_src_notime_results.mp4'
#video_path = 'w_output_video_jet/input_video_12.mp4'




# 動画を読み込む
cap = cv2.VideoCapture(video_path)

# 最初のフレームから5フレームを取得してMatplotlibで表示
for i in range(20):
    ret, frame = cap.read()
    if not ret:
        break
    
    # OpenCVで読み込んだBGR画像をRGBに変換してMatplotlibで表示する準備をする
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    
    # 画像を表示する
    plt.figure(figsize=(8, 6))
    plt.imshow(frame_rgb)
    plt.axis('off')
    #plt.title(f'Frame {i+1}')
    plt.show()

# キャプチャを解放する
cap.release()
```
